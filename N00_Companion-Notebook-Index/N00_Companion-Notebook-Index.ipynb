## How to use the collection

1. Create the shared environment from the repository root with `conda env create -f environment.yml`.
2. Activate it with `conda activate engineering-quantum-interactions`.
3. Open JupyterLab, enter one notebook directory, and run all cells from a fresh kernel.
4. Compare the generated `figures/manuscript/` and `data/` files with the manifest.

All scientific simulations use QuTiP as the quantum-dynamics implementation. NumPy, SciPy, pandas, and Matplotlib support numerical analysis, tabulation, and plotting.

```python
from pathlib import Path
import json, platform, sys
import matplotlib, numpy, pandas, qutip, scipy

HERE = Path.cwd().resolve()
if HERE.name == 'N00_Companion-Notebook-Index':
    repo = HERE.parents[1]
elif (HERE / 'notebooks').is_dir():
    repo = HERE
else:
    raise RuntimeError('Run N00 from its directory or from the repository root.')

manifest = pandas.read_csv(repo / 'notebooks' / 'NOTEBOOK_MANIFEST.csv')
for row in manifest.itertuples(index=False):
    source = repo / row.relative_directory / row.ipynb
    if not source.is_file():
        raise FileNotFoundError(source)

environment = {
    'python': sys.version.split()[0],
    'platform': platform.platform(),
    'qutip': qutip.__version__,
    'numpy': numpy.__version__,
    'scipy': scipy.__version__,
    'pandas': pandas.__version__,
    'matplotlib': matplotlib.__version__,
}
display(manifest[['resource_id', 'title', 'ipynb', 'pdf']])
print(json.dumps(environment, indent=2))
print(f'Validated {len(manifest)} companion notebook sources.')
```
